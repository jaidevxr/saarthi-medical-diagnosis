## 3. Symptom Prevalence Visualization

```python
symptom_cols = [c for c in df_train.columns if c != 'prognosis']
top_symptoms = df_train[symptom_cols].sum().sort_values(ascending=False).head(20)

plt.figure(figsize=(10, 5))
top_symptoms.plot(kind='barh', color='#2b5c8f')
plt.title('Top 20 Most Frequent Symptoms in Dataset', fontsize=12, fontweight='bold')
plt.xlabel('Positive Occurrences')\nplt.gca().invert_yaxis()\nplt.tight_layout()
plt.show()
```
