### Overview
Demonstrates the conversational medical NLP engine:
1. Free-text parsing with Hinglish vocabulary & spelling correction.
2. Symptom mapping to 397 feature indicators.
3. Top-3 differential diagnosis calculation.

```python
import os, sys
PROJECT_ROOT = ".."
sys.path.insert(0, PROJECT_ROOT)

from utils.helpers import parse_symptoms_with_metadata
from models.predictor import DiseasePredictor

predictor = DiseasePredictor(models_dir="../models")

queries = [
    "I have high bukhar, sir dard, and gale me dard since 2 days.",
    "Experiencing chest pain, shortness of breath, and cough.",
    "Pet me dard, vomiting, and nausea."
]

for q in queries:
    res = predictor.predict_from_text(q)
    print(f"\nInput Query:       '{q}'")
    print(f"Detected Symptoms: {res['symptoms_matched']}")
    print(f"Top Diagnosis:     {res['top_prediction']['disease']} ({res['top_prediction']['confidence']}%")
    print(f"Top 3 Candidates:  {[p['disease'] for p in res['top_3']]}")
```
