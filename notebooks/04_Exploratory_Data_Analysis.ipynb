# 📊 04. Exploratory Data Analysis (EDA)
**Project:** Saarthi Medical Diagnosis AI

Visualizes class distribution, symptom frequency distributions, and co-occurrence matrices.

```python
import os, sys, pandas as pd, matplotlib.pyplot as plt
sys.path.insert(0, '..')
from src.utils.constants import PROCESSED_DATA_DIR
df = pd.read_csv(os.path.join(PROCESSED_DATA_DIR, 'Training.csv'))
s_cols = [c for c in df.columns if c != 'prognosis']
top20 = df[s_cols].sum().sort_values(ascending=False).head(20)
plt.figure(figsize=(10, 5))
top20.plot(kind='barh', color='#2b5c8f')
plt.title('Top 20 Symptom Prevalence in Dataset')
plt.gca().invert_yaxis()
plt.show()
```
